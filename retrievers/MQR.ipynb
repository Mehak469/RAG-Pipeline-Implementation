### MQR(Multi-Query Retriever)

```python
from  langchain_classic.retrievers.multi_query import MultiQueryRetriever
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings,ChatGoogleGenerativeAI
from langchain_core.documents import Document
from dotenv import load_dotenv,find_dotenv

load_dotenv(find_dotenv())
```

```python
docs = [
    # --- Programming ---
    Document(page_content="Python is a high-level programming language known for readable syntax, dynamic typing and a huge ecosystem of libraries like NumPy and pandas.",
             metadata={"id": 1, "domain": "programming", "topic": "python-language"}),
    Document(page_content="Java is a statically typed, object-oriented language that runs on the JVM and is widely used for enterprise backends and Android apps.",
             metadata={"id": 2, "domain": "programming", "topic": "java-language"}),
    Document(page_content="Rust is a systems language that guarantees memory safety without a garbage collector, using ownership and borrowing rules.",
             metadata={"id": 3, "domain": "programming", "topic": "rust-language"}),
    Document(page_content="Spring Boot is a Java framework that simplifies building production-ready web services with auto-configuration and embedded servers.",
             metadata={"id": 4, "domain": "programming", "topic": "spring-framework"}),
    Document(page_content="A relational database organizes data into tables of rows and columns, and SQL joins combine tables on shared keys.",
             metadata={"id": 5, "domain": "programming", "topic": "database-table"}),
    Document(page_content="The Unix shell interprets commands such as ls, grep and cd, and lets you chain them together with pipes.",
             metadata={"id": 6, "domain": "programming", "topic": "unix-shell"}),
 
    # --- Animals / nature ---
    Document(page_content="The ball python is a small, docile snake native to West Africa that kills prey by constriction and is popular as a pet.",
             metadata={"id": 7, "domain": "animals", "topic": "python-snake"}),
    Document(page_content="The jaguar is a large cat of the Americas with a powerful bite, rosette-spotted coat, and a love of swimming.",
             metadata={"id": 8, "domain": "animals", "topic": "jaguar-animal"}),
    Document(page_content="A house mouse is a small rodent that nests in walls, breeds quickly and gnaws on wiring and food packaging.",
             metadata={"id": 9, "domain": "animals", "topic": "mouse-animal"}),
    Document(page_content="The whooping crane is a tall North American wading bird that migrates thousands of kilometres between breeding and wintering grounds.",
             metadata={"id": 10, "domain": "animals", "topic": "crane-bird"}),
    Document(page_content="Largemouth bass are freshwater fish that ambush prey near weeds and fallen logs, and are a favourite catch for anglers.",
             metadata={"id": 11, "domain": "animals", "topic": "bass-fish"}),
]
```

```python
embeddings=GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001"
)
```

```python
vector_store=FAISS.from_documents(
    documents=docs,
    embedding=embeddings
)
```

```python
similairty_retriever=vector_store.as_retriever(
    search_type="similarity",
    search_kwargs={"k":2}
)
```

```python
multi_query_retriever=MultiQueryRetriever.from_llm(
    retriever=vector_store.as_retriever(search_kwargs={"k":2}),
    llm=ChatGoogleGenerativeAI(
        model="gemini-3.7-flash"
    )
)
```

```python

query="Is python Slow?"
```

```python
response=similairty_retriever.invoke(query)
response
```
